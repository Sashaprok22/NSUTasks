```python
import csv
import math
import re
from collections import Counter


def assert_raises(exception_type, function, *args, message=None, **kwargs):
    """Проверить тип исключения и, при необходимости, точный текст."""
    try:
        function(*args, **kwargs)
    except exception_type as error:
        if message is not None:
            assert str(error) == message, repr(error)
    else:
        assert False, f"Ожидалось исключение {exception_type.__name__}"
```

## Задача 1. Токенизация текста

Напишите `tokenize(text)`: верните последовательности латинских и кириллических букв (`a–z`, `а–я`, `ё`) в нижнем регистре и в порядке появления. Цифры, апострофы, дефисы и остальные символы разделяют слова. Пустая строка даёт пустой список.


```python
import re

def tokenize(text):
    return re.findall("[a-zа-яё]+", text.lower())
    raise NotImplementedError
```

## Задача 2. Все вхождения подстроки

Реализуйте `naive_search(text, pattern)` перебором начальных позиций: верните индексы всех вхождений, включая перекрывающиеся. Затем напишите `count_substring_occurrences(s, t)`, возвращающую число таких вхождений. Индексы начинаются с нуля. Пустой образец встречается на каждой границе символов: в строке длины `n` таких позиций `n + 1`.


```python
def naive_search(text, pattern):
    result = []
    for i in range(len(text) - len(pattern) + 1):
        if text[i:len(pattern) + i] == pattern:
            result.append(i)
    return result
    raise NotImplementedError

def count_substring_occurrences(s, t):
    return len(naive_search(s, t))
    raise NotImplementedError
```

```python
# Тесты к задаче 2
assert naive_search("abracadabra", "abra") == [0, 7]
assert naive_search("aaaa", "aa") == [0, 1, 2]
assert naive_search("abababa", "aba") == [0, 2, 4]
assert naive_search("коткот", "кот") == [0, 3]
assert naive_search("ab", "abc") == []
assert naive_search("", "a") == []
assert naive_search("ab", "") == [0, 1, 2]
assert naive_search("", "") == [0]
assert count_substring_occurrences("abababa", "aba") == 3
assert count_substring_occurrences("aaaa", "aa") == 3
assert count_substring_occurrences("abc", "z") == 0
assert count_substring_occurrences("", "") == 1
assert count_substring_occurrences("ab", "") == 3
```

## Задача 3. Повторная замена подстрок

Напишите `min_replacement_steps(s, a, b)`. За одну операцию все **неперекрывающиеся** вхождения `a` заменяются на `b`, как при `s.replace(a, b)`; новые вхождения обрабатываются только на следующем шаге. Верните минимальное число операций, после которых `a` исчезнет. Если `a` отсутствует изначально, верните `0`. Если после **1000** операций вхождения остались, вызовите `ValueError("Impossible")`. Успех ровно на 1000-м шаге допустим. Строки состоят из строчных латинских букв; `s` и `b` могут быть пустыми. При пустом `a` вызовите `ValueError`.


```python
def min_replacement_steps(s, a, b):
    if not a: raise ValueError
    
    for i in range(1001):
        if (not(a in s)): break
        s = s.replace(a, b)
    else:
        raise ValueError("Impossible")
    return i
    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError
```

```python
# Тесты к задаче 3
assert min_replacement_steps("abab", "ab", "ba") == 2
assert min_replacement_steps("aaaa", "aa", "a") == 2
assert min_replacement_steps("aaaaa", "aa", "") == 1
assert min_replacement_steps("abc", "x", "y") == 0
assert min_replacement_steps("", "a", "b") == 0
assert min_replacement_steps("a" + "b" * 1000, "ab", "a") == 1000
assert_raises(ValueError, min_replacement_steps, "aaaa", "a", "a", message="Impossible")
assert_raises(ValueError, min_replacement_steps, "a" + "b" * 1001, "ab", "a", message="Impossible")
assert_raises(ValueError, min_replacement_steps, "abc", "", "x")
```

## Задача 4. Тандемные повторы

Тандемный повтор — непустое слово из двух одинаковых половин: `abcabc`, `111111`, `коткот`. Напишите `find_tandems(lines)`, принимающую список строк. Слова выделяйте с помощью `split()` по пробельным символам; пунктуацию не удаляйте, регистр сохраняйте. Верните уникальные тандемные слова **в порядке первого появления** во всех строках.


```python
def find_tandems(lines):
    result = []
    for line in lines:
        words = line.split()
        for word in words:
            if word[:len(word) // 2] == word[len(word) // 2:] and not (word in result):
                result.append(word);
    return result
    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError
```

## Задача 5. RLE: сжатие и восстановление

Напишите `rle_encode(s)` и `rle_decode(encoded)`. Каждую серию одинаковых букв кодируйте как букву и число повторений: `AAABBC → A3B2C1`. Исходные строки содержат только латинские буквы `A–Z`, `a–z`; счётчики могут быть многозначными. Пустая строка кодируется и декодируется в пустую строку. Для некорректного кода (нет числа, число начинается с нуля, посторонний символ) и недопустимых символов исходной строки вызывайте `ValueError`.


```python
def rle_encode(s):
    if re.search("[^a-zA-Z]", s):
            raise ValueError
    last_char = ""
    counter = 0
    result = ""
    for char in s + " ":
        if char == last_char:
            counter += 1
        else:
            if counter > 0:
                result += last_char + str(counter)
            counter = 1
            last_char = char

    return result

    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError

def rle_decode(encoded):
    if re.search("[^a-zA-Z0-9]", encoded):
        raise ValueError

    i = 0
    result = ""
    while i < len(encoded):
        if encoded[i].isdigit(): raise ValueError
        k = i + 1
        repeats = 0
        if k >= len(encoded) or encoded[k] == "0": raise ValueError
        while k < len(encoded) and encoded[k].isdigit():
            repeats = repeats * 10 + int(encoded[k])
            k += 1

        if repeats == 0:
            raise ValueError

        result += encoded[i] * repeats
        i = k

    return result
    raise NotImplementedError
```

```python
# Тесты к задаче 5
assert rle_encode("AAABBC") == "A3B2C1"
assert rle_encode("AAAABBBCCDAA") == "A4B3C2D1A2"
assert rle_encode("aAAa") == "a1A2a1"
assert rle_encode("A" * 12 + "B") == "A12B1"
assert rle_decode("A12B1") == "A" * 12 + "B"
assert rle_decode("A4B3C2D1A2") == "AAAABBBCCDAA"
assert rle_encode("") == rle_decode("") == ""
for sample in ["A", "ABAB", "zzZZzz", "C" * 101]:
    assert rle_decode(rle_encode(sample)) == sample
for bad in ["A", "3A", "A0", "A01", "A-2", "A2!", "A2B"]:
    assert_raises(ValueError, rle_decode, bad)
assert_raises(ValueError, rle_encode, "A2")
```

## Задача 6. Проверка MAC-адреса

Напишите `is_valid_mac(address)` с помощью `re.fullmatch`. Корректный адрес состоит из шести пар шестнадцатеричных цифр (`0–9`, `a–f`, `A–F`). Разделитель — двоеточие или дефис; **во всём адресе он должен быть одинаковым**. Пробелы по краям недопустимы. Возвращайте `True` или `False`.


```python
def is_valid_mac(address):
    return True if re.fullmatch("([a-fA-F0-9]{2}-){5}[a-fA-F0-9]{2}", address) or re.fullmatch("([a-fA-F0-9]{2}:){5}[a-fA-F0-9]{2}", address) else False
    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError
```

## Задача 7. Расстояние Левенштейна

Реализуйте `levenshtein(a, b)` через динамическое программирование. Вставка, удаление и замена одного символа имеют стоимость `1`; совпадение — `0`. Верните минимальную стоимость превращения `a` в `b`. Регистр учитывается, перестановка соседних символов не является отдельной операцией.


```python
def levenshtein(a, b):
    n = len(a)
    m = len(b)

    dp = [[0] * (m + 1) for _ in range(n + 1)]

    for i in range(n + 1):
        dp[i][0] = i

    for j in range(m + 1):
        dp[0][j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = 0 if a[i - 1] == b[j - 1] else 1

            dp[i][j] = min(
                dp[i - 1][j] + 1,
                dp[i][j - 1] + 1,
                dp[i - 1][j - 1] + cost
            )

    return dp[n][m]
    raise NotImplementedError
```

```python
# Тесты к задаче 7
assert levenshtein("kitten", "sitting") == 3
assert levenshtein("flaw", "lawn") == 2
assert levenshtein("кот", "кит") == 1
assert levenshtein("same", "same") == 0
assert levenshtein("", "") == 0
assert levenshtein("", "abc") == 3
assert levenshtein("abc", "") == 3
assert levenshtein("ab", "ba") == 2
assert levenshtein("A", "a") == 1
assert levenshtein("sitting", "kitten") == 3
```

## Задача 8. Z-функция

Напишите `z_function(s)`. Элемент `z[i]` равен длине наибольшего общего префикса всей строки и суффикса с позиции `i`. Положите `z[0] = 0`; для пустой строки верните `[]`. Используйте границы уже найденного Z-блока, чтобы получить линейную сложность.


```python
def z_function(s):
    n = len(s)

    if n == 0:
        return []

    z = [0] * n

    left = 0
    right = 0

    for i in range(1, n):
        if i < right:
            z[i] = min(right - i, z[i - left])

        while i + z[i] < n and s[z[i]] == s[i + z[i]]:
            z[i] += 1

        if i + z[i] > right:
            left = i
            right = i + z[i]

    return z
    raise NotImplementedError
```

```python
# Тесты к задаче 8
assert z_function("aaaaa") == [0, 4, 3, 2, 1]
assert z_function("abacaba") == [0, 0, 1, 0, 3, 0, 1]
assert z_function("abracadabra") == [0, 0, 0, 1, 0, 1, 0, 4, 0, 0, 1]
assert z_function("abc") == [0, 0, 0]
assert z_function("я") == [0]
assert z_function("") == []
```

## Задача 9. Поиск через Z-функцию

Используя `z_function`, реализуйте `z_search(text, pattern)` для поиска всех вхождений, включая перекрывающиеся. Примените Z-функцию к `pattern + separator + text`; выберите разделитель, которого нет ни в образце, ни в тексте (символ `$` может встречаться во входных строках). Формат результата и поведение для пустого образца — как в задаче 2.


```python
def z_search(text, pattern):
    if pattern == "":
        return list(range(len(text) + 1))

    code = 0

    while True:
        separator = chr(code)

        if separator not in text and separator not in pattern:
            break

        code += 1

    combined = pattern + separator + text

    z = z_function(combined)

    result = []

    pattern_len = len(pattern)
    text_start = pattern_len + 1

    for i in range(text_start, len(combined)):
        if z[i] >= pattern_len:
            result.append(i - text_start)

    return result
    raise NotImplementedError
```

```python
# Тесты к задаче 9
assert z_search("abracadabra", "abra") == [0, 7]
assert z_search("aaaa", "aa") == [0, 1, 2]
assert z_search("a$a$a", "$a") == [1, 3]
assert z_search("ab", "") == [0, 1, 2]
assert z_search("", "") == [0]
assert z_search("", "x") == []
assert z_search("ab", "abc") == []
for text, pattern in [("коткот", "кот"), ("abc", "z"), ("abababa", "aba"), ("$$$", "$$")]:
    assert z_search(text, pattern) == naive_search(text, pattern)
```

## Задача 10. Минимальный период строки

Напишите `minimal_period(s)`, возвращающую длину кратчайшего блока, повторением которого можно получить **всю** строку. Используйте Z-функцию. Длина периода должна делить длину строки; если повторяющегося блока нет, ответ равен `len(s)`. Для пустой строки верните `0`.


```python
def minimal_period(s):
    n = len(s)

    if n == 0: return 0

    z = z_function(s)
    for p in range(1, n):
        if n % p == 0 and z[p] >= n - p:
            return p

    return n
    raise NotImplementedError
```

```python
# Тесты к задаче 10
assert minimal_period("abcabcabc") == 3
assert minimal_period("aaaa") == 1
assert minimal_period("abababab") == 2
assert minimal_period("ababa") == 5
assert minimal_period("abcd") == 4
assert minimal_period("коткот") == 3
assert minimal_period("a") == 1
assert minimal_period("") == 0
```

## Задача 11. Эвристический стемминг

Напишите `simple_stem(word)`. Сначала приведите слово к нижнему регистру. Проверьте окончания **в заданном порядке**: `ing`, `ed`, `es`, `s`, `ly`, `ion`, `ment`, `ness`. Удалите первое подходящее окончание, после которого остаётся не менее трёх символов, и сразу верните результат. Если такого окончания нет, верните слово без изменений. Например, `running → runn`: исправлять удвоенные согласные не нужно. Это учебная эвристика, а не полноценный лингвистический алгоритм.


```python
def simple_stem(word):
    word = word.lower()

    suffixes = [
        "ing",
        "ed",
        "es",
        "s",
        "ly",
        "ion",
        "ment",
        "ness"
    ]

    for suffix in suffixes:
        if word.endswith(suffix):
            stem = word[:-len(suffix)]

            if len(stem) >= 3:
                return stem

    return word
    raise NotImplementedError
```

## Задача 12. Частотные и бинарные векторы

Реализуйте `tf_vector(text, vocab)` и `binary_vector(text, vocab)`. Первая функция возвращает список **чисел вхождений**, вторая — список из `0` и `1`, показывающий наличие слов. Используйте `tokenize`; порядок координат совпадает с порядком слов в `vocab`. Слова вне словаря игнорируйте. Словарь — список уникальных слов в нижнем регистре.


```python
def tf_vector(text, vocab):
    tokens = tokenize(text)
    result = []

    for word in vocab:
        result.append(tokens.count(word))

    return result
    raise NotImplementedError


def binary_vector(text, vocab):
    tokens = tokenize(text)
    result = []

    for word in vocab:
        if word in tokens:
            result.append(1)
        else:
            result.append(0)

    return result
    raise NotImplementedError
```

```python
# Тесты к задаче 12
vocab = ["time", "fruit", "flies", "like", "a", "an", "arrow", "banana"]
assert tf_vector("Time flies like an arrow; fruit flies like a banana.", vocab) == [1, 1, 2, 2, 1, 1, 1, 1]
assert binary_vector("like a banana banana", vocab) == [0, 0, 0, 1, 1, 0, 0, 1]
assert tf_vector("BANANA banana unknown", ["banana", "a"]) == [2, 0]
assert binary_vector("BANANA banana unknown", ["banana", "a"]) == [1, 0]
assert tf_vector("", vocab) == [0] * 8
assert binary_vector("", vocab) == [0] * 8
assert tf_vector("some text", []) == []
assert binary_vector("some text", []) == []
```

## Задача 13. Частоты без стоп-слов

Напишите `word_frequencies(text, stop_words)`: токенизируйте текст, удалите стоп-слова и верните словарь частот оставшихся токенов (допустим `Counter`). Стоп-слова заданы коллекцией слов в нижнем регистре. В результате не должно быть ключей с нулевой частотой.


```python
from collections import Counter

def word_frequencies(text, stop_words):
    return Counter(
        token
        for token in tokenize(text)
        if token not in stop_words
    )
    raise NotImplementedError
```

```python
# Тесты к задаче 13
stop_words = {"a", "an", "the", "and", "of", "in", "on", "to", "is", "are"}
text = "The quick brown fox jumps over the lazy dog. A fox and a dog are friends."
assert word_frequencies(text, stop_words) == {
    "quick": 1, "brown": 1, "fox": 2, "jumps": 1, "over": 1,
    "lazy": 1, "dog": 2, "friends": 1,
}
assert word_frequencies("Кот и кот, а ещё пёс.", {"и", "а"}) == {"кот": 2, "ещё": 1, "пёс": 1}
assert word_frequencies("THE and a", stop_words) == {}
assert word_frequencies("", set()) == {}
assert word_frequencies("Hi hi!", set()) == {"hi": 2}
```

## Задача 14. TF-IDF для корпуса

Напишите `tfidf_matrix(documents)`, возвращающую пару `(vocab, matrix)`: отсортированный по алфавиту список уникальных токенов и список строк матрицы (каждая строка — список чисел). Используйте `tokenize`.

Для токена `w` в документе `d`: `TF = count(w, d) / len(tokens(d))`, `IDF = ln(N / df(w))`, где `N` — число всех документов, включая пустые, а `df(w)` — число документов, содержащих `w`. Вес равен `TF * IDF`; сглаживание и нормализацию вектора не применяйте. Для пустого документа все веса нулевые. Для пустого корпуса верните `([], [])`; для корпуса без токенов сохраните по пустой строке матрицы на каждый документ.


```python
import math
from collections import Counter

def tfidf_matrix(documents):
    if not documents: return [], []

    tokenized_docs = [tokenize(doc) for doc in documents]

    vocab = sorted({
        token
        for tokens in tokenized_docs
        for token in tokens
    })

    if not vocab: return [], [[] for _ in documents]

    N = len(documents)
    df = {word: 0 for word in vocab}

    for tokens in tokenized_docs:
        unique_tokens = set(tokens)

        for word in unique_tokens:
            df[word] += 1

    matrix = []

    for tokens in tokenized_docs:
        if not tokens:
            matrix.append([0.0] * len(vocab))
            continue

        counts = Counter(tokens)
        doc_len = len(tokens)

        row = []

        for word in vocab:
            tf = counts[word] / doc_len
            idf = math.log(N / df[word])

            row.append(tf * idf)

        matrix.append(row)

    return vocab, matrix
    raise NotImplementedError
```

```python
# Тесты к задаче 14
vocab, matrix = tfidf_matrix(["the cat cat", "the dog", "bird"])
assert vocab == ["bird", "cat", "dog", "the"]
expected = [
    [0.0, 2 / 3 * math.log(3), 0.0, 1 / 3 * math.log(3 / 2)],
    [0.0, 0.0, 1 / 2 * math.log(3), 1 / 2 * math.log(3 / 2)],
    [math.log(3), 0.0, 0.0, 0.0],
]
assert len(matrix) == len(expected)
for actual, target in zip(matrix, expected):
    assert len(actual) == len(vocab)
    assert all(math.isclose(x, y, rel_tol=1e-9, abs_tol=1e-12) for x, y in zip(actual, target))
assert tfidf_matrix([]) == ([], [])
assert tfidf_matrix(["", "123"]) == ([], [[], []])
assert tfidf_matrix(["кот кот", "КОТ"]) == (["кот"], [[0.0], [0.0]])
vocab, matrix = tfidf_matrix(["cat", ""])
assert vocab == ["cat"]
assert len(matrix) == 2 and all(len(row) == 1 for row in matrix)
assert math.isclose(matrix[0][0], math.log(2))
assert matrix[1] == [0.0]
```

## Задача 15. Символьные и словесные N-граммы

Реализуйте `char_ngrams(s, n)` и `word_ngrams(text, n)`. Первая возвращает список подстрок длины `n`, сохраняя регистр и пробелы. Вторая применяет `tokenize` и возвращает список **кортежей** из `n` соседних токенов. Перекрытия и повторы сохраняются. Если элементов меньше `n`, верните `[]`. Для целого `n <= 0` вызовите `ValueError`. Постройте через `Counter` частоты словесных биграмм.


```python
from collections import Counter

def char_ngrams(s, n):
    if n <= 0: raise ValueError("n must be positive")
    if len(s) < n: return []

    return [
        s[i:i + n]
        for i in range(len(s) - n + 1)
    ]
    raise NotImplementedError


def word_ngrams(text, n):
    if n <= 0: raise ValueError("n must be positive")
    tokens = tokenize(text)
    if len(tokens) < n: return []

    return [
        tuple(tokens[i:i + n])
        for i in range(len(tokens) - n + 1)
    ]
    raise NotImplementedError
```

```python
# Тесты к задаче 15
assert char_ngrams("ababa", 2) == ["ab", "ba", "ab", "ba"]
assert char_ngrams("A б", 2) == ["A ", " б"]
assert char_ngrams("abc", 3) == ["abc"]
assert char_ngrams("ab", 3) == []
assert char_ngrams("", 1) == []
assert word_ngrams("Кот и кот, и пёс.", 2) == [("кот", "и"), ("и", "кот"), ("кот", "и"), ("и", "пёс")]
assert word_ngrams("One two THREE", 3) == [("one", "two", "three")]
assert word_ngrams("Hi!", 1) == [("hi",)]
assert word_ngrams("one", 2) == []
assert word_ngrams("", 1) == []
assert Counter(word_ngrams("кот и кот и пёс", 2)) == {("кот", "и"): 2, ("и", "кот"): 1, ("и", "пёс"): 1}
for n in [0, -1]:
    assert_raises(ValueError, char_ngrams, "abc", n)
    assert_raises(ValueError, word_ngrams, "a b", n)
```

## Задача 16. Поиск Бойера—Мура

Реализуйте упрощённый поиск с правилом плохого символа. Функция `build_last_table(pattern)` возвращает словарь `символ → индекс последнего вхождения` (индексы с нуля). Функция `boyer_moore(text, pattern)` сравнивает символы справа налево. При несовпадении в позиции `j` образца с символом `c` текста сдвиг равен `max(1, j - last.get(c, -1))`. После полного совпадения можно сдвигаться на один символ. Верните **все** позиции, включая перекрывающиеся; пустой образец обработайте как в задаче 2. Эвристику хорошего суффикса реализовывать не нужно.


```python
def build_last_table(pattern):
    last = {}

    for i in range(len(pattern)):
        last[pattern[i]] = i

    return last
    raise NotImplementedError


def boyer_moore(text, pattern):
    if pattern == "":
        return list(range(len(text) + 1))

    last = build_last_table(pattern)

    result = []

    n = len(text)
    m = len(pattern)

    shift = 0

    while shift <= n - m:
        j = m - 1

        while j >= 0 and pattern[j] == text[shift + j]: j -= 1

        if j < 0:
            result.append(shift)
            shift += 1

        else:
            c = text[shift + j]
            shift += max(
                1,
                j - last.get(c, -1)
            )

    return result
    raise NotImplementedError
```

```python
# Тесты к задаче 16
assert build_last_table("abaca") == {"a": 4, "b": 1, "c": 3}
assert build_last_table("") == {}
assert boyer_moore("test test test", "test") == [0, 5, 10]
assert boyer_moore("aaaa", "aa") == [0, 1, 2]
assert boyer_moore("abracadabra", "abra") == [0, 7]
assert boyer_moore("ab", "") == [0, 1, 2]
assert boyer_moore("", "") == [0]
assert boyer_moore("", "a") == []
assert boyer_moore("ab", "abc") == []
# Полный перебор коротких строк проверяет пропуски и перекрытия.
from itertools import product
short_strings = ["".join(chars) for n in range(5) for chars in product("ab", repeat=n)]
for text in short_strings:
    for pattern in short_strings:
        assert boyer_moore(text, pattern) == naive_search(text, pattern)
```

## Задача 17. Классификация спама по мешку слов

Напишите `classify_message(text, spam_docs, ham_docs)`. Токенизируйте обучающие документы и суммируйте частоты токенов отдельно в спаме и обычных сообщениях. Для каждого токена нового текста добавьте к оценке `spam_count[token] - ham_count[token]`, учитывая повторы в новом тексте. Верните `"spam"`, если сумма строго положительна; иначе `"ham"`. Неизвестные токены дают нулевой вклад.


```python
from collections import Counter

def classify_message(text, spam_docs, ham_docs):
    spam_count = Counter()
    ham_count = Counter()

    for doc in spam_docs: spam_count.update(tokenize(doc))
    for doc in ham_docs: ham_count.update(tokenize(doc))

    score = 0

    for token in tokenize(text):
        score += spam_count[token] - ham_count[token]

    if score > 0: return "spam"
    return "ham"
    raise NotImplementedError
```

```python
# Тесты к задаче 17
spam_docs = ["Buy buy now", "free prize"]
ham_docs = ["meeting now", "project meeting"]
assert classify_message("BUY prize!", spam_docs, ham_docs) == "spam"
assert classify_message("project meeting", spam_docs, ham_docs) == "ham"
assert classify_message("now", spam_docs, ham_docs) == "ham"
assert classify_message("buy meeting", spam_docs, ham_docs) == "ham"
assert classify_message("buy buy meeting", spam_docs, ham_docs) == "spam"
assert classify_message("unknown", spam_docs, ham_docs) == "ham"
assert classify_message("", spam_docs, ham_docs) == "ham"
assert classify_message("buy", [], []) == "ham"
```

## Задача 18. Конвейер обработки текста и сходство

Реализуйте `text_similarity(a, b, stop_words)`. Для каждого текста выполните `tokenize → simple_stem → удаление стоп-слов → Counter`. Здесь стоп-слова уже заданы в нижнем регистре **после стемминга**. Вычислите косинусную близость частотных векторов по объединению слов двух текстов: скалярное произведение, делённое на произведение евклидовых норм. Если хотя бы один вектор нулевой, верните `0.0`.


```python
from collections import Counter
import math


def text_similarity(a, b, stop_words):
    def process(text):
        tokens = tokenize(text)

        stems = [
            simple_stem(token)
            for token in tokens
        ]

        filtered = [
            stem
            for stem in stems
            if stem not in stop_words
        ]

        return Counter(filtered)

    counter_a = process(a)
    counter_b = process(b)

    words = set(counter_a) | set(counter_b)

    dot_product = sum(
        counter_a[word] * counter_b[word]
        for word in words
    )

    norm_a = math.sqrt(sum(
        counter_a[word] ** 2
        for word in words
    ))

    norm_b = math.sqrt(sum(
        counter_b[word] ** 2
        for word in words
    ))

    if norm_a == 0 or norm_b == 0: return 0.0

    return dot_product / (norm_a * norm_b)
    raise NotImplementedError
```

```python
# Тесты к задаче 18
assert math.isclose(text_similarity("Cats played", "PLAYED cats", set()), 1.0)
assert math.isclose(text_similarity("cats played", "cat play", set()), 1.0)
assert math.isclose(text_similarity("cat cat dog", "cat dog dog", set()), 0.8)
assert math.isclose(text_similarity("the cat", "cat", {"the"}), 1.0)
assert math.isclose(text_similarity("cats dog", "dog", {"cat"}), 1.0)
assert text_similarity("cat", "airplane", set()) == 0.0
assert text_similarity("", "cat", set()) == 0.0
assert text_similarity("", "", set()) == 0.0
assert text_similarity("the and", "cat", {"the", "and"}) == 0.0
assert math.isclose(text_similarity("a a b", "a b b", set()), text_similarity("a b b", "a a b", set()))
```

Сдайте ноутбук с решением, сравнением и выводами, а также полученный файл `news_keywords.csv`.


```python
import csv
import math
from collections import Counter


def analyze_news(input_path, output_path):
    with open(input_path, "r", encoding="utf-8", newline="") as file:
        reader = csv.DictReader(file)
        rows = list(reader)

    if not rows:
        with open(output_path, "w", encoding="utf-8", newline=""):
            pass
        return

    documents = [
        tokenize(row["text"])
        for row in rows
    ]

    N = len(documents)
    df = Counter()

    for tokens in documents: df.update(set(tokens))

    with open(output_path, "w", encoding="utf-8", newline="") as file:
        writer = csv.writer(file)

        for row, tokens in zip(rows, documents):
            counts = Counter(tokens)

            by_frequency = sorted(
                counts,
                key=lambda word: (-counts[word], word)
            )

            by_frequency = by_frequency[:10]
            tfidf = {}

            if tokens:
                doc_len = len(tokens)

                for word in counts:
                    tf = counts[word] / doc_len
                    idf = math.log(N / df[word])

                    tfidf[word] = tf * idf

            by_tfidf = sorted(
                counts,
                key=lambda word: (-tfidf[word], word)
            )

            by_tfidf = by_tfidf[:10]

            writer.writerow([
                row["title"],
                ",".join(by_frequency),
                ",".join(by_tfidf)
            ])

analyze_news("news.csv", "news_keywords.csv")
```
