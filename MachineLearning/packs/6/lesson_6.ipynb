## Задача 2. Все вхождения подстроки

Реализуйте `naive_search(text, pattern)` перебором начальных позиций: верните индексы всех вхождений, включая перекрывающиеся. Затем напишите `count_substring_occurrences(s, t)`, возвращающую число таких вхождений. Индексы начинаются с нуля. Пустой образец встречается на каждой границе символов: в строке длины `n` таких позиций `n + 1`.


```python
def naive_search(text, pattern):
    result = []
    for i in range(len(text) - len(pattern) + 1):
        if text[i:len(pattern) + i] == pattern:
            result.append(i)
    return result
    raise NotImplementedError

def count_substring_occurrences(s, t):
    return len(naive_search(s, t))
    raise NotImplementedError
```

```python
# Тесты к задаче 2
assert naive_search("abracadabra", "abra") == [0, 7]
assert naive_search("aaaa", "aa") == [0, 1, 2]
assert naive_search("abababa", "aba") == [0, 2, 4]
assert naive_search("коткот", "кот") == [0, 3]
assert naive_search("ab", "abc") == []
assert naive_search("", "a") == []
assert naive_search("ab", "") == [0, 1, 2]
assert naive_search("", "") == [0]
assert count_substring_occurrences("abababa", "aba") == 3
assert count_substring_occurrences("aaaa", "aa") == 3
assert count_substring_occurrences("abc", "z") == 0
assert count_substring_occurrences("", "") == 1
assert count_substring_occurrences("ab", "") == 3
```

```python
def min_replacement_steps(s, a, b):
    if not a: raise ValueError
    
    for i in range(1001):
        if (not(a in s)): break
        s = s.replace(a, b)
    else:
        raise ValueError("Impossible")
    return i
    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError
```

## Задача 4. Тандемные повторы

Тандемный повтор — непустое слово из двух одинаковых половин: `abcabc`, `111111`, `коткот`. Напишите `find_tandems(lines)`, принимающую список строк. Слова выделяйте с помощью `split()` по пробельным символам; пунктуацию не удаляйте, регистр сохраняйте. Верните уникальные тандемные слова **в порядке первого появления** во всех строках.


```python
def find_tandems(lines):
    result = []
    for line in lines:
        words = line.split()
        for word in words:
            if word[:len(word) // 2] == word[len(word) // 2:] and not (word in result):
                result.append(word);
    return result
    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError
```

## Задача 5. RLE: сжатие и восстановление

Напишите `rle_encode(s)` и `rle_decode(encoded)`. Каждую серию одинаковых букв кодируйте как букву и число повторений: `AAABBC → A3B2C1`. Исходные строки содержат только латинские буквы `A–Z`, `a–z`; счётчики могут быть многозначными. Пустая строка кодируется и декодируется в пустую строку. Для некорректного кода (нет числа, число начинается с нуля, посторонний символ) и недопустимых символов исходной строки вызывайте `ValueError`.


```python
def rle_encode(s):
    if re.search("[^a-zA-Z]", s):
            raise ValueError
    last_char = ""
    counter = 0
    result = ""
    for char in s + " ":
        if char == last_char:
            counter += 1
        else:
            if counter > 0:
                result += last_char + str(counter)
            counter = 1
            last_char = char

    return result

    # ВАШ КОД ЗДЕСЬ
    raise NotImplementedError

def rle_decode(encoded):
    if re.search("[^a-zA-Z0-9]", encoded):
        raise ValueError

    i = 0
    result = ""
    while i < len(encoded):
        if encoded[i].isdigit(): raise ValueError
        k = i + 1
        repeats = 0
        if k >= len(encoded) or encoded[k] == "0": raise ValueError
        while k < len(encoded) and encoded[k].isdigit():
            repeats = repeats * 10 + int(encoded[k])
            k += 1

        if repeats == 0:
            raise ValueError

        result += encoded[i] * repeats
        i = k

    return result
    raise NotImplementedError
```

```python
# Тесты к задаче 5
assert rle_encode("AAABBC") == "A3B2C1"
assert rle_encode("AAAABBBCCDAA") == "A4B3C2D1A2"
assert rle_encode("aAAa") == "a1A2a1"
assert rle_encode("A" * 12 + "B") == "A12B1"
assert rle_decode("A12B1") == "A" * 12 + "B"
assert rle_decode("A4B3C2D1A2") == "AAAABBBCCDAA"
assert rle_encode("") == rle_decode("") == ""
for sample in ["A", "ABAB", "zzZZzz", "C" * 101]:
    assert rle_decode(rle_encode(sample)) == sample
for bad in ["A", "3A", "A0", "A01", "A-2", "A2!", "A2B"]:
    assert_raises(ValueError, rle_decode, bad)
assert_raises(ValueError, rle_encode, "A2")
```
