### Задание 32. Категории происшествий

Получите частоты всех типов ДТП из столбца `category`. Определите наиболее распространённый тип и число различных типов.

### Задание 33. Наезды на пешеходов без уличного освещения

Подсчитайте ДТП, у которых `category` равен **«Наезд на пешехода»**, а `light` равен **«В темное время суток, освещение отсутствует»**. Для фильтра используйте две логические маски.


```python
pedestrian_mask = accidents['category'].eq('Наезд на пешехода')
no_lighting_mask = accidents['light'].eq('В темное время суток, освещение отсутствует')
print("Число ДТП:", (pedestrian_mask & no_lighting_mask).sum())
```

### Задание 34. Дневные и ночные ДТП

Найдите доли ДТП с погибшими: среди дневных и среди ночных, с округлением до сотых. **Сумерки не включаются** ни в день, ни в ночь.



```python
day_mask = accidents['light'].eq('Светлое время суток')
night_mask = accidents['light'].str.startswith('В темное время суток', na=False)
fatal_mask = accidents['dead_count'] > 0
fatal_shares = pd.Series({
    'День': fatal_mask.loc[day_mask].mean(),
    'Ночь': fatal_mask.loc[night_mask].mean(),
}, name='Доля ДТП с погибшими').round(2)
display(fatal_shares)
print("Сумерки и неустановленное освещение исключены; доли указаны от 0 до 1.")
```

### Задание 35. В какие дни недели происходит больше всего ДТП?

1. Преобразуйте столбец `datetime` в тип даты и времени. Строки содержат фрагмент вида `Thu Jan 01 2015 06:30:00 GMT+0600 (...)`; для корректного преобразования сначала уберите часть, начинающуюся с ` GMT`.
2. Создайте столбец `weekday`: номер дня недели от 0 (понедельник) до 6 (воскресенье).
3. Подсчитайте количество ДТП по дням недели и выведите результат в календарном порядке: понедельник, ..., воскресенье.
4. Определите дни с максимальным и минимальным числом ДТП.
5. **Дополнительно:** найдите среднее число ДТП на один календарный день отдельно для будней и выходных. Для корректного сравнения учитывайте количество реальных дат каждого типа в наблюдаемом периоде (оно необязательно одинаковое).

**Подсказки:** `str.replace(..., regex=True)`, `pd.to_datetime`, `.dt.dayofweek`, `value_counts`, `sort_index`, `groupby`. Если `datetime` содержит пропуски, предусмотрите `errors="coerce"` при преобразовании. Сравнивайте **количество зарегистрированных ДТП**, а не дорожный риск на одну поездку.


```python
datetime_text = accidents['datetime'].str.replace(r' GMT.*$', '', regex=True)
accidents['datetime'] = pd.to_datetime(
    datetime_text, format='%a %b %d %Y %H:%M:%S', errors='coerce'
)
accidents['weekday'] = accidents['datetime'].dt.day_of_week
weekday_names = pd.Index(['Понедельник', 'Вторник', 'Среда', 'Четверг',
                          'Пятница', 'Суббота', 'Воскресенье'])
weekday_counts = accidents['weekday'].value_counts().reindex(range(7), fill_value=0).sort_index()
weekday_counts.index = weekday_names
display(weekday_counts.rename('Число ДТП'))
print("Максимум:", weekday_counts[weekday_counts.eq(weekday_counts.max())].index.tolist())
print("Минимум:", weekday_counts[weekday_counts.eq(weekday_counts.min())].index.tolist())
print("Дат не распознано:", accidents['datetime'].isna().sum())

calendar_days = pd.date_range(accidents['datetime'].min().normalize(),
                              accidents['datetime'].max().normalize(), freq='D')
daily_counts = accidents.groupby(accidents['datetime'].dt.normalize()).size().reindex(calendar_days, fill_value=0)
day_types = np.where(daily_counts.index.day_of_week < 5, 'Будни', 'Выходные')
weekday_weekend = daily_counts.groupby(day_types).agg(['sum', 'size', 'mean'])
weekday_weekend.columns = ['Число ДТП', 'Календарных дней', 'ДТП на один день']
display(weekday_weekend)
print("Период:", calendar_days.min().date(), "—", calendar_days.max().date())
print("Это частота зарегистрированных ДТП, а не дорожный риск на одну поездку.")
```
