Напишите функцию `divide(a, b)`: в `try` — вычисление `a / b`; в `except ZeroDivisionError` — вывод «деление на ноль» и возврат `None`; в блоке `else` — вывод «успешно» и возврат результата. Убедитесь, что блок `else` выполняется **только** при успехе.

````python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("деление на ноль")
        return None
    else:
        print("успешно")
        return result

divide(1, 0)
divide(4, 2)
````

Напишите функцию-генератор `countdown(n)`, которая с помощью `while` и `yield` последовательно отдаёт числа от `n` до `1`. Выведите её элементы циклом `for`, затем превратите в список через `list()`.


````python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for n in countdown(5):
    print(n)

lst = list(countdown(5))
````

Напишите генератор `evens(limit)`, который отдаёт чётные числа от 0 до `limit` (не включая). Создайте генератор `g = evens(6)` и получите первые три значения функцией `next(g)`.


````python
def evens(limit):
    for n in range(0, limit, 2):
        yield n

gen = evens(6)

print(next(gen), next(gen), next(gen))
````

Напишите декоратор `echo(func)`, внутренняя функция которого добавляет в список `log` метки `'start'` и `'end'` вокруг вызова исходной функции. Примените его к функции `hello()` через `@echo` и выведите `log`.


````python
log = []

def echo(func):
    def decorated(*args, **kwargs):
        log.append("start")
        func(*args, **kwargs)
        log.append("end")
    return decorated

@echo
def hello():
    print('Hello!')

hello()
````

Напишите функцию `positive_sqrt(x)`, которая возвращает квадратный корень из числа `x`.

Если `x` отрицательное, функция должна самостоятельно вызвать исключение `ValueError` с помощью `raise` и сообщением `"отрицательное число"`.

Проверьте работу функции на двух значениях:
- положительном числе;

Вызов функции для отрицательного числа поместите в `try/except` и обработайте возникшее исключение, выведя его сообщение.

````python
def positive_sqrt(n):
    if n < 0:
        raise ValueError("отрицательное число")
    else:
        return n ** 0.5

print(positive_sqrt(25))

try:
    print(positive_sqrt(-25))
except ValueError as e:
    print(e)
````

Напишите бесконечный генератор `fib()`, который последовательно возвращает числа Фибоначчи:

`0, 1, 1, 2, 3, 5, 8, ...`

Для возврата очередного числа используйте `yield`. Генератор не должен заранее ограничивать количество выдаваемых значений.

Создайте генератор и с помощью функции `next()` получите из него первые 10 чисел Фибоначчи. Выведите каждое полученное число.

````python
def fib():
    n1 = 0
    n2 = 0
    while True:
        yield n1 + n2
        if (n1 == 0 and n2 == 0):
            n1 = 1
            continue
        temp = n1
        n1 = n2
        n2 = temp + n2

gen = fib()

for _ in range(10):
    print(next(gen))
````

Напишите генератор `squares(limit)`, отдающий квадраты чисел от 1 до `limit`. Получите значения функцией `next()` в цикле `while True`, а когда генератор исчерпается — перехватите `StopIteration`, выведите «генератор исчерпан» и выйдите циклом `break`.


````python
def squares(limit):
    for n in range(1, limit + 1):
        yield n ** 2

gen = squares(10)
try:
    while True:
        print(next(gen))
except StopIteration:
    print("генератор исчерпан")
````

Напишите декоратор `log_call`, который перед вызовом функции выводит:

- имя вызываемой функции;
- переданные позиционные аргументы `args`;
- переданные именованные аргументы `kwargs`.

Внутренняя функция декоратора должна принимать произвольное количество аргументов с помощью `*args` и `**kwargs` и передавать их исходной функции без изменений.

Примените декоратор к функции calculate_price, которая вычисляет итоговую стоимость покупки с учётом скидки в процентах.

Вызовите функцию несколько раз:
- только с позиционными аргументами;
- с именованными аргументами;
- одновременно с позиционными и именованными аргументами.

Для каждого вызова выведите переданные аргументы и полученный результат.

````python
def log_call(func):
    def decorated(*args, **kwargs):
        print(func.__name__, args, kwargs)
        return func(*args, **kwargs)
    return decorated


@log_call
def calculate_price(price, count, discount=0):
    total = price * count
    return total * (1 - discount / 100)


calculate_price(10, 5)
calculate_price(10, 5, discount=8)
````

Напишите функцию-генератор `averager()`, которая используется как **корутина**: она получает числа по одному и после каждого нового числа возвращает среднее арифметическое всех полученных значений.

Для получения нового значения внутри корутины используйте конструкцию:

```python
value = yield avg
```

где `avg` — текущее среднее.

Перед первой передачей числа запустите корутину с помощью `next()`. Затем передавайте значения методом `send()`.

Проверьте работу корутины на последовательности `10`, `20`, `30`. После каждого вызова `send()` должно возвращаться текущее среднее:

- после `10` → `10.0`;
- после `20` → `15.0`;
- после `30` → `20.0`.

````python
def average():
    count = 0
    total = 0

    while True:
        if (count == 0):
            x = yield 0
        else:
            x = yield total / count
        count += 1
        total += x

gen = average()
next(gen)
gen.send(5)
gen.send(4)
print(gen.send(3))
````
